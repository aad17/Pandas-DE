```python
import pandas as pd
```

```python
claims = pd.DataFrame({
    "claim_id": ["C101", "C102", "C103", "C104", "C105", "C106", "C107", "C108"],
    "hospital": ["Apollo", "Ruby", "Apollo", "Sahyadri", "Ruby", "Apollo", "Sahyadri", "Ruby"],
    "claim_type": ["Surgery", "Emergency", "OPD", "Surgery", "Emergency", "Surgery", "OPD", "Emergency"],
    "claim_amount": [85000, 42000, 5000, 125000, 67000, 35000, 8000, 95000],
    "status": ["Approved", "Pending", "Approved", "Pending", "Rejected", "Pending", "Approved", "Pending"]
})
```

Find claims where:

claim_amount > 50000
AND status == "Pending"

```python
claims[(claims['claim_amount'] > 50000) & (claims['status'] == 'Pending')]
```

Find all claims whose claim_type is either:

Surgery \
Emergency

Use isin().

```python
claims[claims['claim_type'].isin(['Surgery', 'Emergency'])]
```

Find claims with claim_amount between 30,000 and 90,000 inclusive

```python
claims[claims['claim_amount'].between(30000, 90000)]
```

Using loc[], return only:

claim_id, hospital, claim_amount \
for claims with claim_amount > 60000

```python
claims.loc[claims['claim_amount'] > 60000, ['claim_id', 'hospital', 'claim_amount']]
```

Sort the entire dataset by claim_amount, highest first.

```python
claims.sort_values('claim_amount', ascending=False)
```

Return the 3 largest claims by claim_amount

```python
claims.nlargest(n=3, columns='claim_amount')
```
