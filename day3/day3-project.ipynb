### Business Requirements

Build the investigation output satisfying all of these rules:

1. Claim status must be Pending.
2. Claim type must be either Surgery or Emergency.
3. Claim amount must be between ₹70,000 and ₹200,000 inclusive.
4. Patient age must be at least 40.
5. Final output should contain only:
claim_id, hospital, claim_type, claim_amount, patient_age
6. Sort suspicious claims by claim_amount, highest first.
7. Separately, find the 2 highest-value claims in the entire original dataset, regardless of whether they're suspicious.

```python
df = claims[
    (claims['status'] == 'Pending') & 
    (claims['claim_type'].isin(['Surgery', 'Emergency'])) &
    (claims['claim_amount'].between(70000, 200000)) &
    (claims['patient_age'] >= 40)
].loc[:, ['claim_id', 'hospital', 'claim_type', 'claim_amount', 'patient_age']] \
.sort_values(by='claim_amount', ascending=False)
```
