```python
import pandas as pd
```

```python
data = {
    "Txn_ID": ["T101", "T102", "T102", "T103", "T104"],
    "Customer_ID": [101, 102, 102, None, 104],
    "Amount": ["500.0", "750.0", "750.0", "1200.0", "300.0"],
    "Status": [" success ", "FAILED", "FAILED", None, " Success"]
}
```

```python
df = pd.DataFrame(data)
```

Investigation

```python
df.head()
```

```python
df.info()
```

Find the number of null values in every column.

```python
df.isna().sum()
```

Find duplicate transactions based on Txn_ID.

```python
df[df.duplicated(subset=['Txn_ID'])]
```

Remove the duplicate T102.

```python
df.drop_duplicates(subset=['Txn_ID'], inplace=True)
```

```python
df
```

Rename Txn_ID → transaction_id

```python
df.rename(columns={
    'Txn_ID': 'transaction_id'
}, inplace=True)
```

```python
df
```

Convert Amount to float

```python
df['Amount'] = df['Amount'].astype("float64")
```

Standardize Status so values such as " success " and " Success" become "SUCCESS"

```python
df['Status'] = df['Status'].str.strip().str.upper()
```

Replace missing status values with "UNKNOWN".

```python
df['Status'] = df['Status'].fillna(value="UNKNOWN")
```

```python
df
```

Remove records where Customer_ID is missing

```python
df.dropna(subset=["Customer_ID"], inplace=True)
```

```python
df
```

```python
df.info()
```
