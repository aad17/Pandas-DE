```python
import pandas as pd
```

```python
events = pd.DataFrame({
    "account_id": [101, 101, 102, 102, 102, 103],
    "status": [
        "ACTIVE", "BLOCKED",
        "ACTIVE", "SUSPENDED", "ACTIVE",
        "ACTIVE"
    ],
    "updated_at": pd.to_datetime([
        "2026-09-20 09:00",
        "2026-09-20 11:00",
        "2026-09-20 08:00",
        "2026-09-20 12:00",
        "2026-09-20 12:00",
        "2026-09-20 10:00"
    ]),
    "event_sequence": [1, 2, 1, 2, 3, 1]
})
```

Sort the events by account_id and then updated_at.

```python
sorted_df = events.sort_values(
    ['account_id', 'updated_at']
)
```

```python
sorted_df
```

Using sort + drop_duplicates(), return one latest row per account. Account 102 has a timestamp tie, so use event_sequence as the tie-breaker.

```python
sorted_events = events.sort_values(
    ['account_id', 'updated_at', 'event_sequence']
)
```

```python
sorted_events.drop_duplicates(
    subset=['account_id'],
    keep='last'
)
```

Using groupby().idxmax(), return the row having the maximum updated_at for each account.

```python
idx = sorted_events.groupby('account_id')['updated_at'].idxmax()
events.loc[idx]
```
