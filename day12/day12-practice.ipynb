```python
import pandas as pd
```

Exercise 1 - Concat

```python
old = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "city": ["Pune", "Mumbai", "Delhi"]
})

new = pd.DataFrame({
    "customer_id": [104, 105],
    "city": ["Chennai", "Bengaluru"]
})
```

```python
pd.concat([old, new], ignore_index=True)
```

Exercise 3 — Anti-join

```python
target = pd.DataFrame({
    "order_id": [1001, 1002, 1003],
})

batch = pd.DataFrame({
    "order_id": [1002, 1004, 1005],
})
```

```python
joined = batch.merge(
    target,
    how='left',
    on='order_id',
    indicator=True
)
```

```python
joined[joined['_merge'] == 'left_only']
```
