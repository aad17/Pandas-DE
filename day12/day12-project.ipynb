### Part F - Derived Columns

```python
joined
```

```python
joined = joined[joined['_merge'] != 'left_only']
```

```python
joined['order_date'] = joined['order_ts'].dt.date
```

```python
def get_order_size(amount):
    if amount >= 15000:
        return 'High'
    elif amount >= 5000 and amount < 15000:
        return 'Medium'
    elif amount < 5000:
        return 'Low'

joined['order_size'] = joined['amount'].apply(get_order_size)
```

### Part G - Regional KPI Table

```python
regional_kpi = joined.groupby('region').agg(
    total_orders=('order_id', 'count'),
    unique_customers=('customer_id', 'nunique'),
    total_revenue=('amount', 'sum'),
    avg_order_value=('amount', 'mean'),
    max_order_value=('amount', 'max')
)
```

```python
regional_kpi
```

```python
regional_kpi = regional_kpi.sort_values(by='total_revenue', ascending=False)
```

```python
regional_kpi
```

### Part H - Customer Level Analytics

```python
customer_analytics = joined.groupby(['customer_id', 'customer_name', 'region']).agg(
    total_orders=('order_id', 'count'),
    total_spend=('amount', 'sum'),
    avg_order_value=('amount', 'mean')
)
```

```python
customer_analytics
```

```python
customer_analytics["regional_avg_customer_spend"] = customer_analytics.groupby('region')['total_spend'].transform('mean')
```

```python
customer_analytics
```

```python
customer_analytics['spend_vs_regional_avg'] = customer_analytics['total_spend'] - customer_analytics['regional_avg_customer_spend']
```

```python
sorted_joined = joined.sort_values(['customer_id', 'order_ts'])
```

```python
sorted_joined['previous_order_amount'] = sorted_joined.groupby('customer_id')['amount'].shift(1)
```

```python
sorted_joined['amount_change'] = sorted_joined.groupby('customer_id')['amount'].diff()
```

```python
sorted_joined['cumulative_spend'] = sorted_joined.groupby('customer_id')['amount'].cumsum()
```
