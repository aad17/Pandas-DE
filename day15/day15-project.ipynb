## Clean

```python
orders['order_status'] = orders['order_status'].str.strip().str.upper()
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
```

```python
customers['region'] = customers['region'].str.strip().str.capitalize()
```

```python
customers['updated_at'] = pd.to_datetime(customers['updated_at'])
```

```python
payments['payment_status'] = payments['payment_status'].str.strip().str.upper()
```

## Validate

```python
valid_orders = orders[
    ~(orders['customer_id'].isna()) & 
    (orders['amount'] > 0) & 
    (orders['order_status'].isin(['COMPLETED', 'CANCELLED']))
]
```

```python
invalid_orders = orders[~orders['order_id'].isin(valid_orders['order_id'])]
```

## Reconcile Payments

```python
completed_orders = valid_orders[valid_orders['order_status'] == 'COMPLETED']
```

```python
joined = valid_orders.merge(
    payments,
    how='outer',
    on='order_id',
    indicator=True,
    validate='1:1'
)
```

```python
orphan_successful_payments = joined[(joined['payment_status'] == 'SUCCESS') & (joined['_merge'] == 'right_only')]
```

```python
completed_orders_without_successful_payments = joined[
    (joined['order_status'] == 'COMPLETED') & 
    (joined['payment_status'] != 'SUCCESS')
]
```

## Build curated other dataset

```python
df = completed_orders_with_successful_payments.merge(
    latest_records,
    how='inner',
    on='customer_id'
)
```

```python
df = df.loc[
    :,
    [
        'order_id',
        'order_date',
        'customer_id',
        'customer_name',
        'region',
        'amount',
        'payment_amount'
    ]
]
```

```python
df
```

```python
df['revenue_match'] = df['amount'] == df['payment_amount']
```

## Daily regional summary

```python
daily_regional_summary = df.groupby([df['order_date'].dt.date, 'region']).agg(
    successful_orders=('order_id', 'count'),
    unique_customers=('customer_id', 'nunique'),
    total_order_amount=('amount', 'sum'),
    total_payment_amount=('payment_amount', 'sum')
)
```

```python
daily_regional_summary = daily_regional_summary.sort_values(
    ['order_date', 'region']
)
```
