```python
import pandas as pd
```

```python
data = {
    "shipment_id": ["S1", "S2", "S3", "S4", "S5", "S6", "S7"],
    "warehouse": ["Pune", "Pune", "Mumbai", "Mumbai", "Pune", "Delhi", "Delhi"],
    "region": ["West", "West", "West", "West", "West", "North", "North"],
    "customer_id": ["C1", "C2", "C1", "C3", "C1", "C4", "C4"],
    "revenue": [1200, 1800, 2500, 900, 1500, 3000, 2000]
}

df = pd.DataFrame(data)
```

```python
df
```

Calculate total revenue by warehouse.

```python
df.groupby('warehouse', as_index=False).agg(
    total_revenue=('revenue', 'sum')
)
```

Calculate average revenue by region.

```python
df.groupby('region', as_index=False).agg(
    avg_revenue=('revenue', 'mean')
)
```

```python
df.groupby('warehouse', as_index=False).agg(
    unique_customers=('customer_id', 'nunique')
)
```

Produce one warehouse-level table containing:

- total_revenue
- avg_revenue
- max_revenue
- unique_customers

```python
df.groupby('warehouse', as_index=False).agg(
    total_revenue=('revenue', 'sum'),
    avg_revenue=('revenue', 'mean'),
    max_revenue=('revenue', 'max'),
    unique_customers=('customer_id', 'nunique')
)
```

Produce revenue grouped by region + warehouse.

```python
df.groupby(['region', 'warehouse'], as_index=False).agg(
    total_revenue=('revenue', 'sum')
)
```
