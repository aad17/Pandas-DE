```python
import pandas as pd
```

```python
sales = pd.DataFrame({
    "region": ["East", "East", "East", "West", "West", "West"],
    "month": ["Jan", "Jan", "Feb", "Jan", "Feb", "Feb"],
    "category": ["Laptop", "Phone", "Laptop", "Laptop", "Phone", "Laptop"],
    "revenue": [50000, 30000, 60000, 40000, 35000, 45000]
})
```

```python
sales.head()
```

Create a table with region as rows, month as columns, and total revenue as values.

```python
region_revenue = sales.pivot_table(
    index='region',
    columns='month',
    values='revenue',
    aggfunc='sum'
)
```

```python
region_revenue
```

Create a table with region as rows, category as columns, and total revenue as values.

```python
category_revenue = sales.pivot_table(
    index='region',
    columns='category',
    values='revenue',
    aggfunc='sum'
)
```

```python
category_revenue
```

Take your result from #1 and use melt() to convert it back toward long format. Keep region as the identifier.

```python
region_revenue.reset_index().melt(
    id_vars='region',
    var_name='month',
    value_name='revenue'
)
```

Using groupby(), create total revenue by region + category, then use unstack() to turn category into columns.

```python
grouped = sales.groupby(['region', 'category']).agg(
    total_revenue=('revenue', 'sum')
)
```

```python
grouped
```

```python
grouped.unstack()
```
