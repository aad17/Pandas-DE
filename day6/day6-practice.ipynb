```python
import pandas as pd
```

```python
data = {
    "device_id": ["D101", "D102", "D103", "D101", "D104"],
    "event_time": [
        "2026-09-20 08:15:00",
        "2026-09-20 10:30:00",
        "invalid",
        "2026-09-20 15:45:00",
        "2026-09-21 01:10:00"
    ],
    "temperature": [42, 55, 48, 67, 39]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

Convert event_time to datetime. Invalid timestamps should become missing datetime values.

```python
df['event_time'] = pd.to_datetime(df['event_time'], errors="coerce")
```

```python
df
```

Find all records whose timestamp failed conversion.

```python
df[df['event_time'].isna()]
```

Create an event_hour column.

```python
df['event_hour'] = df['event_time'].dt.hour
```

```python
df
```

```python
df['event_date'] = df['event_time'].dt.date
```

```python
df
```

Return events occurring from 10:00 inclusive to 16:00 exclusive on September 20.

```python
df.loc[
    (df['event_hour'] >= 10) & 
    (df['event_hour'] < 16) &
    (df['event_date'].astype("str") == "2026-09-20"),
    :
]
```

Create window_end = event_time + 30 minutes

```python
df['window_end'] = df['event_time'] + pd.Timedelta(minutes=30)
```
