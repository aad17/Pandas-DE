```python
products = pd.json_normalize(orders_json, record_path='items', meta=['order_id'])
```

```python
products
```

```python
products['item_revenue'] = products['quantity'] * products['unit_price']
```

```python
products
```

```python
exploded_df = exploded_df.loc[
    :,
    [
        'order_id',
        'order_date',
        'status',
        'customer_customer_id',
        'customer_name',
        'customer_city',
        'shipping_method',
        'shipping_warehouse_id'
    ]
]
```

```python
products
```

```python
exploded_df = exploded_df.drop_duplicates()
```

```python
exploded_df
```

```python
final_df = exploded_df.merge(
    products,
    on='order_id',
    how='inner'
)
```

```python
final_df
```

1. How many unique orders were received?

```python
final_df['order_id'].nunique()
```

2. How many item records exist after flattening? \
-> 6

3. What is the total revenue from DELIVERED orders only?

```python
final_df.groupby('status').agg(
    total_revenue=('item_revenue', 'sum')
).loc['DELIVERED']
```

4. Calculate delivered revenue by customer_city.

```python
final_df[final_df['status'] == 'DELIVERED'].groupby('customer_city').agg(
    total_revenue=('item_revenue', 'sum')
)
```

5. Which product_id generated the highest delivered revenue?

```python
final_df[final_df['status'] == 'DELIVERED'].sort_values('item_revenue', ascending=False).iloc[0]
```
