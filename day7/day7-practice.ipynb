```python
import pandas as pd
import numpy as np
```

```python
transactions = pd.DataFrame({
    "transaction_id": [101, 102, 103, 104, 105, 106],
    "customer_id": ["C1", "C2", "C3", "C1", "C4", "C5"],
    "amount": [1200, 75000, 42000, 110000, 800, 65000],
    "payment_method": ["UPI", "CARD", "CARD", "UPI", "CASH", "CARD"],
    "country": ["IN", "IN", "US", "SG", "IN", "US"]
})
```

```python
transactions.head()
```

Create high_value_flag:

- "Y" when amount >= 50000
- "N" otherwise
- Use np.where().

```python
transactions['high_value'] = np.where(
    transactions['amount'] >= 50000,
    "Y",
    "N"
)
```

```python
transactions
```

Create international_flag:

- "Y" when country != "IN"
- "N" otherwise.

```python
transactions['international_flag'] = np.where(
    transactions['country'] != "IN",
    "Y",
    "N"
)
```

```python
transactions
```

Given:
payment_risk = {
    "UPI": "LOW",
    "CARD": "MEDIUM",
    "CASH": "HIGH"
}

Create payment_risk using map().

```python
payment_risk = {
    "UPI": "LOW",
    "CARD": "MEDIUM",
    "CASH": "HIGH"
}
```

```python
transactions['payment_risk'] = transactions['payment_method'].map(payment_risk)
```

```python
transactions
```

Create amount_with_fee using a vectorized operation:
- amount_with_fee = amount * 1.02

```python
transactions['amount_with_fee'] = transactions['amount'] * 1.02
```

```python
transactions
```

Create review_flag:
- REVIEW if:
  - amount >= 50000
  - AND
  - country != "IN"

otherwise:
- OK

```python
transactions['review_flag'] = np.where(
    (transactions['amount'] >= 50000) & 
    (transactions['country'] != 'IN'),
    "REVIEW",
    "OK"
)
```

```python
transactions
```
