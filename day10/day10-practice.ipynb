```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "region": ["West", "West", "West", "East", "East", "East"],
    "salesperson": ["A", "B", "C", "D", "E", "F"],
    "revenue": [500, 800, 500, 900, 600, 600],
    "orders": [10, 15, 8, 20, 12, 14]
})
```

```python
df.head()
```

Create:
- revenue_rank

- Rank salespeople by revenue within their region, highest revenue first.
- Tied revenue should receive the same rank and the next rank should not have a gap.

```python
df['revenue_rank'] = df.groupby('region')['revenue'].rank(method='dense', ascending=False)
```

```python
df.sort_values(['region', 'revenue_rank'])
```

Create:
- region_avg_revenue

- containing the average revenue of that salesperson's region while keeping all six original rows.

```python
df['region_avg_revenue'] = df.groupby('region')['revenue'].transform('mean')
```

```python
df
```

Using the column created in Q2, create:
- above_region_avg

- which is True when the salesperson's revenue is greater than their region average.

```python
df['above_region_revenue'] = df['revenue'] > df['region_avg_revenue']
```

```python
df
```

Create:
- region_total_revenue
- revenue_share

- where revenue_share represents each salesperson's contribution to their region's total revenue.
- For example, 0.25 means 25% of regional revenue.

```python
df['region_total_revenue'] = df.groupby('region')['revenue'].transform('sum')
```

```python
df
```

```python
df['revenue_share'] = df['revenue'] / df['region_total_revenue'] * 100
```
