```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [
        "O101", "O102", "O103", "O104",
        "O105", "O106", "O107", "O108",
        "O109", "O110", "O111", "O112"
    ],
    
    "seller_id": [
        "S01", "S02", "S01", "S03",
        "S03", "S04", "S05", "S04",
        "S05", "S06", "S02", "S06"
    ],
    
    "category": [
        "Electronics", "Electronics", "Electronics", "Electronics",
        "Electronics", "Furniture", "Furniture", "Furniture",
        "Furniture", "Furniture", "Electronics", "Furniture"
    ],
    
    "order_date": [
        "2026-08-02", "2026-08-04", "2026-08-08", "2026-08-10",
        "2026-08-15", "2026-08-03", "2026-08-06", "2026-08-12",
        "2026-08-18", "2026-08-20", "2026-07-29", "2026-08-22"
    ],
    
    "amount": [
        1200, 1800, 800, 2000,
        2000, 1500, 1000, 500,
        1000, 700, 5000, 300
    ],
    
    "status": [
        "Delivered", "Delivered", "Delivered", "Cancelled",
        "Delivered", "Delivered", "Delivered", "Delivered",
        "Delivered", "Delivered", "Delivered", "Cancelled"
    ]
})
```

Converting order date to datatime

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
```

```python
orders.info()
```

```python
filtered_orders = orders.loc[
    (orders['order_date'] >= '2026-08-01') & 
    (orders['order_date'] < '2026-09-01') &
    (orders['status'] == 'Delivered'), 
    :
] 
```

```python
filtered_orders
```

```python
grouped = filtered_orders.groupby(['seller_id', 'category'], as_index=False).agg(
    total_revenue=('amount', 'sum'),
    order_count=('order_id', 'count')
)
```

```python
grouped
```

```python
grouped['category_avg_revenue'] = grouped.groupby('category')['total_revenue'].transform('mean')
```

```python
grouped
```

```python
grouped['revenue_vs_category_avg'] = grouped['total_revenue'] - grouped['category_avg_revenue']
```

```python
grouped
```

```python
grouped['category_total_revenue'] = grouped.groupby('category')['total_revenue'].transform('sum')
```

```python
grouped
```

```python
grouped['category_revenue_share'] = grouped['total_revenue'] / grouped['category_total_revenue']
```

```python
grouped
```

```python
grouped['revenue_rank'] = grouped.groupby('category')['total_revenue'].rank(method='dense', ascending=False)
```

```python
grouped
```

```python
grouped.sort_values(['category', 'revenue_rank', 'seller_id'])
```
